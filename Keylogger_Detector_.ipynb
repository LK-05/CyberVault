```python
import psutil
import time

# Define suspicious keywords typically found in keylogger programs
suspicious_keywords = ["keylogger", "keystroke", "logger", "capture", "record"]

def check_process(process):
    """Checks if a process name or description matches suspicious keywords."""
    try:
        process_name = process.name().lower()
        process_cmdline = ' '.join(process.cmdline()).lower()

        # Look for keywords in the process name or command line
        for keyword in suspicious_keywords:
            if keyword in process_name or keyword in process_cmdline:
                return True
    except (psutil.NoSuchProcess, psutil.AccessDenied):
        pass
    return False

def detect_keyloggers():
    """Continuously scans for suspicious processes that may indicate keyloggers."""
    print("Starting keylogger detection...")
    while True:
        suspicious_processes = []
        for process in psutil.process_iter():
            if check_process(process):
                suspicious_processes.append((process.pid, process.name()))

        if suspicious_processes:
            print("Suspicious processes detected:")
            for pid, name in suspicious_processes:
                print(f"Process ID: {pid}, Process Name: {name}")
        else:
            print("No suspicious processes detected.")

        # Scan every 10 seconds
        time.sleep(10)

# Run the detector
detect_keyloggers()
```
