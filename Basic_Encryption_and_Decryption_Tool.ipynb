1. Caesar Cipher (Shift Cipher)

```python
# Caesar Cipher
def caesar_encrypt(plaintext, shift):
    encrypted_text = ""
    for char in plaintext:
        if char.isalpha():
            shift_base = ord('a') if char.islower() else ord('A')
            encrypted_text += chr((ord(char) - shift_base + shift) % 26 + shift_base)
        else:
            encrypted_text += char
    return encrypted_text

def caesar_decrypt(encrypted_text, shift):
    decrypted_text = ""
    for char in encrypted_text:
        if char.isalpha():
            shift_base = ord('a') if char.islower() else ord('A')
            decrypted_text += chr((ord(char) - shift_base - shift) % 26 + shift_base)
        else:
            decrypted_text += char
    return decrypted_text

# Test Caesar Cipher
shift = 3
plaintext = input("Enter text to encrypt (Caesar Cipher): ")
encrypted_text = caesar_encrypt(plaintext, shift)
decrypted_text = caesar_decrypt(encrypted_text, shift)

print(f"Encrypted Text: {encrypted_text}")
print(f"Decrypted Text: {decrypted_text}")
```

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
import base64

def pad(text):
    # Pad text to be multiple of 16 bytes
    return text + (16 - len(text) % 16) * chr(16 - len(text) % 16)

def unpad(text):
    # Remove padding
    return text[:-ord(text[len(text)-1:])]

def aes_encrypt(plaintext, key):
    # Encrypt using AES
    iv = get_random_bytes(16)  # Random IV for security
    cipher = AES.new(key, AES.MODE_CBC, iv)
    padded_text = pad(plaintext)
    encrypted_bytes = cipher.encrypt(padded_text.encode('utf-8'))
    encrypted_text = base64.b64encode(iv + encrypted_bytes).decode('utf-8')
    return encrypted_text

def aes_decrypt(encrypted_text, key):
    # Decrypt using AES
    encrypted_bytes = base64.b64decode(encrypted_text)
    iv = encrypted_bytes[:16]  # Extract IV from the beginning
    encrypted_text_bytes = encrypted_bytes[16:]
    cipher = AES.new(key, AES.MODE_CBC, iv)
    decrypted_text = unpad(cipher.decrypt(encrypted_text_bytes).decode('utf-8'))
    return decrypted_text

# Test AES Encryption
key = b'Sixteen byte key'  # 16-byte key for AES-128 (must be 16, 24, or 32 bytes)
plaintext = input("Enter text to encrypt (AES): ")
encrypted_text = aes_encrypt(plaintext, key)
decrypted_text = aes_decrypt(encrypted_text, key)

print(f"Encrypted Text: {encrypted_text}")
print(f"Decrypted Text: {decrypted_text}")
```
