```python
import hashlib
import secrets

def hash_password(password: str) -> str:
    # Generate a random salt
    salt = secrets.token_hex(16)

    # Combine salt with password and hash it
    salted_password = salt + password
    hashed_password = hashlib.sha256(salted_password.encode()).hexdigest()

    # Store salt with the hashed password (needed for password verification)
    return f"{salt}${hashed_password}"

def verify_password(stored_password: str, password_attempt: str) -> bool:
    # Split stored password into salt and hash
    salt, hashed_password = stored_password.split('$')

    # Hash the attempt with the stored salt
    salted_attempt = salt + password_attempt
    attempt_hash = hashlib.sha256(salted_attempt.encode()).hexdigest()

    # Compare the attempt hash with the stored hash
    return attempt_hash == hashed_password

# Example Usage
# Hashing a new password
password = "my_secure_password"
stored_password = hash_password(password)
print("Stored password hash:", stored_password)

# Verifying the password
attempt = input("Enter your password to verify: ")
if verify_password(stored_password, attempt):
    print("Password verified!")
else:
    print("Password does not match.")
```
