```python
# Importing necessary libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
# Define function to scrape URLs
def scrape_urls():
  # Placeholder for scraping URLs from PhishTank or similar sites
    phishing_urls = [
        # Example URLs - replace with scraping code if allowed or manual dataset
        "http://example-phishing.com",
        "http://fakebank-login.com",
        "http://safe-site.org"  # Legitimate URL
    ]
    return phishing_urls
```

```python
# Feature extraction function
def extract_features(url):
    features = {}
    features['url_length'] = len(url)
    features['contains_https'] = 1 if 'https' in url else 0
    features['num_dots'] = url.count('.')
    features['num_hyphens'] = url.count('-')
    # Add more features as needed
    return features
```

```python
# Prepare dataset
def prepare_data(urls, labels):
    data = [extract_features(url) for url in urls]
    df = pd.DataFrame(data)
    df['label'] = labels
    return df
```

```python
# Load data
phishing_urls = scrape_urls()
# Labels: 1 for phishing, 0 for legitimate
labels = [1, 1, 0]
```

```python
# Prepare data
data = prepare_data(phishing_urls, labels)
```

```python
# Split data into training and testing sets
X = data.drop('label', axis=1)
y = data['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Initialize and train model
model = RandomForestClassifier()
model.fit(X_train, y_train)
```

```python
# Predict and evaluate
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# Predict on new URLs
new_urls = ["http://new-phish-example.com", "http://legit-website.com"]
new_data = pd.DataFrame([extract_features(url) for url in new_urls])
predictions = model.predict(new_data)
print("Predictions:", predictions)  # 1 for phishing, 0 for legitimate
```
